In order to code an automatic way of making predictions, let's generalize our network. A general network would take $n$ inputs, would have many hidden layers, each hidden layer having $m$ nodes, and would have an output layer. Although the network is showing one hidden layer, but we will code the network to have many hidden layers. Similarly, although the network shows an output layer with one node, we will code the network to have more than one node in the output layer.<img src="http://cocl.us/general_neural_network" alt="Neural Network General" width="600px">



Build a Neural Network


```python
n = 2 # no of inputs
num_hidden_layers = 2 # no of hidden layers
m = [2,2] #no of nodes in each hidden layer
num_nodes_output_layer = 1 # no of nodes in the output layer
```

```python
import numpy as np
num_nodes_previous_layer = n # number of nodes in the previous layer
network = {} # initialize the network as an empty dictionary

for layer in range(num_hidden_layers + 1): # iterate through the number of hidden layers
    if layer == num_hidden_layers:
        layer_name = "output"
        num_nodes = num_nodes_output_layer
    else:
        layer_name = 'layer_{}'. format(layer + 1 )
        num_nodes = m[layer]

# initialize the weights and biases for the current layer
network[layer_name] = {}
for node in range(num_nodes):
    node_name = 'node_{}'.format(node + 1)
    network[layer_name][node_name] = {
            'weights': np.around(np.random.uniform(size=num_nodes_previous_layer), decimals=2),
            'bias': np.around(np.random.uniform(size=1), decimals=2),
        }
    
    num_nodes_previous = num_nodes
    
print(network) # print network
```

```python
def initialise_nwtwork(num_inputs, num_hidden_layers, num_nodes_hidden, num_nodes_ouput):
    num_nodes_previous = num_inputs # number of nodes in the previous layer
    network = {} # initialize the network as an empty dictionary

    #loop through each layer and initialize the weights and biases for each node
    for layer in range(num_hidden_layers + 1): # iterate through the number of hidden
        if layer == num_hidden_layers:
            layer_name = "output"
            num_nodes = num_nodes_ouput
        else:
            layer_name = 'layer_{}'. format(layer + 1 )
            num_nodes = num_nodes_hidden[layer] 

        #initialise weights and biases for the each node
        network[layer_name] = {}
        # for node in range(num_nodes):
        #     node_name = 'node_{}'.format(node + 1)
        #     network[layer_name][node_name] = {
        #             'weights': np.around(np.random.uniform(size=num_nodes_previous), decimals=2),
        #             'bias': np.around(np.random.uniform(size=1), decimals=2),
        #         }
            
        #     num_nodes_previous = num_nodes
        for node in range(num_nodes):
            node_name = 'node_{}'.format(node + 1)
            network[layer_name][node_name] = {
                    'weights': np.around(np.random.uniform(size=num_nodes_previous), decimals=2),
                    'bias': np.around(np.random.uniform(size=1), decimals=2),
                }
            
        num_nodes_previous = num_nodes
    return network # return the initialized network

    
```

## Compute Weighted Sum at Each Node


```python
def compute_weighted_sum(inputs, weights, bias):
    return np.sum(inputs * weights) + bias
```

Generate 5 input that we can feed to small_network

```python
from random import seed
import numpy as np

np.random.seed(12)
inputs = np.around(np.random.uniform(size=5), decimals=2)

print('The inputs to the network are {}'.format(inputs)) # print the inputs to the network
```

Use the compute_weighted_sum function to compute the weighted sum at the first node in the first hidden layer.

```python
weights_l1_n1 = small_network['layer_1']['node_1']['weights']
bias_l1_n1 = small_network['layer_1']['node_1']['bias']
z_11_l1 = compute_weighted_sum(inputs, weights_l1_n1, bias_l1_n1)
print('The weighted sum at the first node in the first hidden layer is {}'.format(np.around(z_11_l1, decimals=4)))
```

Compute Node Activation

```python
def node_activation(weighted_sum):
    return 1.0 / (1.0 + np.exp(-weighted_sum))
```

Use the node_activation function to compute the output of the first node in the first hidden layer.

```python
a_11_l1 = node_activation(z_11_l1)
print('The activation of the first node in the first hidden layer is {}'.format(np.around(a_11_l1, decimals=4)))
```

Forward Propagation


The final piece of building a neural network that can perform predictions is to put everything together. So let's create a function that applies the compute_weighted_sum and node_activation functions to each node in the network and propagates the data all the way to the output layer and outputs a prediction for each node in the output layer.

The way we are going to accomplish this is through the following procedure:

Start with the input layer as the input to the first hidden layer.
Compute the weighted sum at the nodes of the current layer.
Compute the output of the nodes of the current layer.
Set the output of the current layer to be the input to the next layer.
Move to the next layer in the network.
Repeat steps 2 - 5 until we compute the output of the output layer.

```python
def forward_propagate(network, inputs):

    layer_inputs = list(inputs) # convert the inputs to a list

    for layer in network:
        layer_data = network[layer] # get the data for the current layer
        layer_outputs = []
        for layer_node in layer_data:
            node_data = layer_data[layer_node] # get the data for the current node
            node_output = node_activation(compute_weighted_sum(layer_inputs, node_data['weights'], node_data['bias'])) # compute the output of the current node
            layer_outputs.append(np.around(node_output[0], decimals=4)) # append the output of the current node to the list of layer outputs
        if layer != 'output':
            print('The activations of the nodes in {} are {}'.format(layer, layer_outputs)) # print the activations of the nodes in the current layer
        layer_inputs = layer_outputs
    network_prediction = layer_outputs
    print('The output of the network is {}'.format(network_prediction)) # print the output of the network
    return network_prediction
```

Use the forward_propagate function to compute the prediction of our small network
